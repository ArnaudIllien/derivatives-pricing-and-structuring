```python
#-Fonction pour la date
from datetime import date

def maturite_en_annees(date_calcul: date, date_fin: date):
    """
    Retourne la maturité en fraction d'année (jours / 365).
    """
    if date_fin < date_calcul:
        raise ValueError("La date de fin doit être postérieure ou égale à la date de calcul.")
    return (date_fin - date_calcul).days/365


#-Paramètres-

options="Call"
date_calcul = date(2025, 9, 1)   # aujourd'hui
date_fin = date(2025, 9, 30)      # maturité 
spot = 200                       # prix du sous-jacent
strike = 190                     # prix d'exercice
tx_sansrisque = 0.04                      # taux sans risque (4%)
vol = 0.36                      #volatilité

#-Constante-

from scipy.stats import norm
import math 
def call_option(date_calcul, date_fin, spot, strike, tx_sansrisque, vol):
    temps=maturite_en_annees(date_calcul, date_fin)
    if temps == 0:
        return max(spot - strike, 0)   # à maturité : valeur intrinsèque
    D1= (math.log(spot/strike) + (tx_sansrisque + (vol**2)/2)*temps)/((temps**(1/2))*vol)
    D2=D1-(temps**(1/2))*vol
    C=spot*norm.cdf(D1) - math.exp(-tx_sansrisque*temps)*strike*norm.cdf(D2)
    
    return C

def put_option(date_calcul, date_fin, spot, strike, tx_sansrisque, vol):
    temps=maturite_en_annees(date_calcul, date_fin)
    if temps == 0:
        return max(strike - spot, 0)   # à maturité : valeur intrinsèque
    D1= (math.log(spot/strike) + (tx_sansrisque + (vol**2)/2)*temps)/((temps**(1/2))*vol)
    D2=D1-(temps**(1/2))*vol
    P= math.exp(-tx_sansrisque*temps)*strike*norm.cdf(-D2) - spot*norm.cdf(-D1)
    
    return P

#- Pricing -

if options =="Call":
    pricing_option=call_option(date_calcul, date_fin, spot, strike, tx_sansrisque, vol)
    print("Prix du call", pricing_option)
else :
    pricing_option=put_option(date_calcul, date_fin, spot, strike, tx_sansrisque, vol)
    print("Prix du put", pricing_option)
```
